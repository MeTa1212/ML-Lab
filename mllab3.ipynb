```python
import numpy as np
import pandas as pd
data=pd.read_csv("enjoy.csv")
```

```python
x=data.iloc[:,:-1].values
y=data.iloc[:,-1].values
```

```python
def find_s(x,y):
  hypothesis=None
  print("Initial Hypothesis: ",hypothesis)
  for i in range(len(x)):
    if y[i]=="Yes":
      if hypothesis is None:
        hypothesis=x[i].copy()
      else:
        for j in range(len(hypothesis)):
          if hypothesis[j]!=x[i][j]:
            hypothesis[j]="?"
    print(f"\n After Training example {i+1}")
    print(hypothesis)
  return hypothesis
final_hypothesis=find_s(x,y)
print("\n Final Hypothesis: ")
print(final_hypothesis)
```

```python
def candidate_elemination(concepts,target):
  specific=concepts[0].copy()
  general=[["?" for i in range(len(specific))]]
  print("\n Initial S: ",specific)
  print("\n Initial G: ",general)
  for i,h in enumerate(concepts):
    if target[i]=="Yes":
      for x in range(len(specific)):
        if h[x]!=specific[x]:
          specific[x]="?"
    else:
      for x in range(len(specific)):
        if h[x]!=specific[x]:
          general.append(["?" if j!=x else specific[x] for j in range(len(specific))])
    print("\n Example",i+1)
    print("S=",specific)
    print("G=",general)
  return specific,general
S,G=candidate_elemination(x,y)
print("\n Final specific Boundary")
print(S)
print("\n Final General Boundary")
print(G)
```
